```python
##download/load DistilBERT
!mkdir -p /content/distilbert-base-uncased

!wget -O /content/distilbert-base-uncased/config.json \
https://huggingface.co/distilbert/distilbert-base-uncased/resolve/main/config.json

!wget -O /content/distilbert-base-uncased/vocab.txt \
https://huggingface.co/distilbert/distilbert-base-uncased/resolve/main/vocab.txt

!wget -O /content/distilbert-base-uncased/pytorch_model.bin \
https://huggingface.co/distilbert/distilbert-base-uncased/resolve/main/pytorch_model.bin
```

```python
from pathlib import Path

path = Path("/content/Reproducibility/bary_score_re.py")
text = path.read_text()

text = text.replace(
    'load_name = "/content/bert-base-uncased" if self.model_name == "bert-base-uncased" else self.model_name',
    '''if self.model_name == "bert-base-uncased":
                load_name = "/content/bert-base-uncased"
            elif self.model_name == "distilbert-base-uncased":
                load_name = "/content/distilbert-base-uncased"
            else:
                load_name = self.model_name'''
)

path.write_text(text)

print("BaryScore now supports DistilBERT")
```
