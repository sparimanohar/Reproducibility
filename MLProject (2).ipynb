**BertScore on WMT18**

```python
from pathlib import Path

path = Path("/content/Reproducibility/bert_score/utils.py")
text = path.read_text()

text = text.replace(
    "model = AutoModel.from_pretrained(model_type)",
    'load_name = "/content/bert-base-uncased" if model_type == "bert-base-uncased" else model_type\n        model = AutoModel.from_pretrained(load_name)'
)

text = text.replace(
    "tokenizer = AutoTokenizer.from_pretrained(model_type)",
    'load_name = "/content/bert-base-uncased" if model_type == "bert-base-uncased" else model_type\n        tokenizer = AutoTokenizer.from_pretrained(load_name)'
)

path.write_text(text)
print("patched")
```

**BaryScore on WMT18 dataset**

```python
from pathlib import Path

path = Path("/content/Reproducibility/bary_score_re.py")
text = path.read_text()

text = text.replace(
    "tokenizer = AutoTokenizer.from_pretrained('{}'.format(self.model_name))",
    'load_name = "/content/bert-base-uncased" if self.model_name == "bert-base-uncased" else self.model_name\n            tokenizer = AutoTokenizer.from_pretrained(load_name)'
)

text = text.replace(
    "model = AutoModelForMaskedLM.from_pretrained('{}'.format(self.model_name))",
    "model = AutoModelForMaskedLM.from_pretrained(load_name)"
)

path.write_text(text)

print("BaryScore model loading patched")
```
